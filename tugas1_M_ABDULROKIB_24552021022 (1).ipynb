````python
# Membuat file input.txt berisi angka 1-10
with open('input.txt', 'w') as f:
    for i in range(1, 11):
        f.write(f"{i}\n")

print("File input.txt berhasil dibuat!")
````

````python
# ============================================
# Nama  : M.ABDUL ROKIB
# NIM   : 24552021022
# Tugas : Praktikum Google Colab
# ============================================

print("=== Hasil Output ===")
with open('input.txt', 'r') as f:
    for baris in f:
        angka = int(baris.strip())

        # Ganjil → kali 2, Genap → tampilkan apa adanya
        if angka % 2 != 0:
            print(angka * 2)
        else:
            print(angka)
````

````python
from google.colab import userdata
import requests

# Pastikan Anda telah menambahkan 'github_token' di menu Secrets (ikon kunci di sidebar kiri)
try:
    token = userdata.get('github_token')
except Exception as e:
    print("⚠️ Gagal mengambil token dari Secrets Colab. Pastikan Anda telah membuat secret bernama 'github_token'")
    token = None

if token:
    r = requests.get(
        "https://api.github.com/user",
        headers={"Authorization": f"token {token}"}
    )

    if r.status_code == 200:
        print("✅ Token valid. Login sebagai:", r.json()['login'])
    else:
        print("❌ Token tidak valid. Status:", r.status_code)
        print(r.json())
````
